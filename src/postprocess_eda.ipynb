```python
import pandas as pd
import re

gold = pd.read_csv('/data/ephemeral/home/nlp-5/eunbyul/ah/data/dev.csv')

# 길이, 문장 수 분포
summary_lengths = gold['summary'].astype(str).str.len()
summary_sents = gold['summary'].astype(str).apply(lambda x: len(re.split(r'[.!?]', x))-1)

print(f"평균 길이: {summary_lengths.mean():.1f}, 95%: {summary_lengths.quantile(0.95):.1f}")
print(f"평균 문장 수: {summary_sents.mean():.2f}, 95%: {summary_sents.quantile(0.95)}")

# 대표 패턴/토큰 빈도
all_summaries = ' '.join(gold['summary'].astype(str))
print("스페셜 토큰:", dict(pd.Series(re.findall(r"#\w+#", all_summaries)).value_counts().head(10)))
print("자주 쓰는 동사:", pd.Series(re.findall(r"(요청|제안|설명|논의|전달|축하|조언|권유|결정|공유|수락|거절|추천|안내|토론)", all_summaries)).value_counts().head(10))
```
